```python
import pandas as pd
from autoviz.AutoViz_Class import AutoViz_Class
import sweetviz as sv
```

```python
import pandas as pd

df = pd.read_csv("/content/train.csv")
df.head()
```

```python
AV = AutoViz_Class()

AV.AutoViz(
    filename="/content/train.csv",
    sep=",",
    depVar="",
    dfte=None,
    header=0,
    verbose=2
)
```

```python
import sweetviz as sv

report = sv.analyze(df)

report.show_html("sweetviz_report.html")
```

Insights

1. The Technology category generated the highest sales.
2. The West region recorded the highest overall sales.
3. Most sales values are concentrated at lower amounts, with a few high-value orders.
